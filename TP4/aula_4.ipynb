```python
from python_files.ImageHough import HoughPlane, ShowHoughLines, ShowHoughLineSegments, ShowHoughCircles, ShowVideo
import matplotlib.pyplot as plt
import cv2 as cv
import numpy as np

img_orig = cv.imread("Aula_4_files/aula4-3.bmp")
img_orig_gray = cv.imread("Aula_4_files/aula4-3.bmp", cv.IMREAD_GRAYSCALE)
```

Ex 1

```python
img = img_orig_gray.copy()
acc, theta, rho = HoughPlane(img, -90, 90, 1)


plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.axis('off')
plt.subplot(1, 2, 2)
plt.imshow(acc, cmap="gray")
plt.axis('off')
plt.show()
```

Ex 2

```python
img = img_orig.copy()
img_gray = img_orig_gray.copy()

img_line = ShowHoughLines(img_gray, img, 4)

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.axis('off')
plt.subplot(1, 2, 2)
plt.imshow(img_line)
plt.axis('off')
plt.show()
```

Ex 3

```python

img_orig = cv.imread("Aula_4_files/aula4-2.bmp")
img_orig_gray = cv.cvtColor(img_orig, cv.COLOR_BGR2GRAY)

imgSobel = cv.Sobel(img_orig_gray, cv.CV_8U, 1, 0)
cv.threshold(imgSobel, 255, 255, cv.THRESH_OTSU + cv.THRESH_BINARY, imgSobel)

img_line = ShowHoughLines(imgSobel, img_orig, 120)
img_line = cv.cvtColor(img_line, cv.COLOR_BGR2RGB)

plt.subplot(1, 2, 1)
plt.imshow(imgSobel, cmap='gray')
plt.axis('off')
plt.subplot(1, 2, 2)
plt.imshow(img_line)
plt.axis('off')
plt.show()
```

Ex 4

```python
img_orig = cv.imread("Aula_4_files/aula4-2.bmp")
img_orig_gray = cv.imread("Aula_4_files/aula4-2.bmp", cv.IMREAD_GRAYSCALE)

imgCanny = cv.Canny(img_orig_gray, 150, 100)

img_line_p = ShowHoughLineSegments(imgCanny, img_orig, 20, 10, 5)
img_line_p = cv.cvtColor(img_line_p, cv.COLOR_BGR2RGB)

plt.imshow(imgCanny, cmap='gray')
plt.axis('off')
plt.show()
plt.imshow(img_line_p)
plt.axis('off')
plt.show()
```

```python
img_orig = cv.imread("Aula_4_files/aula4-coins.png")
img_orig_gray = cv.imread("Aula_4_files/aula4-coins.png", cv.IMREAD_GRAYSCALE)

img_circles = ShowHoughCircles(img_orig_gray, img_orig, 100, 100, 50, 80, 100, 3)
img_circles = cv.cvtColor(img_circles, cv.COLOR_BGR2RGB)

plt.imshow(img_orig_gray, cmap='gray')
plt.axis('off')
plt.show()
plt.imshow(img_circles)
plt.axis('off')
plt.show()
```

Ex 6

```python
cap = cv.VideoCapture("Aula_4_files/aula4-video.mp4")

while cap.isOpened():
    ret, frame = cap.read()

    # if frame is read correctly ret is True
    if not ret:
        print("Can't receive frame (stream end?). Exiting ...")
        break

    imgCanny = cv.Canny(frame, 100, 70)

    img_line_p = ShowHoughLineSegments(imgCanny, frame, 20, 100, 5)

    cv.imshow('frame', img_line_p)
    if cv.waitKey(1) == ord('q'):
        break

cv.destroyAllWindows()
```
